```python
#importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

```python
#load the datasets.
df = sns.load_dataset("iris")
df.head()
```

```python
sns.pairplot(df)
```

```python
df.drop(columns=['species'],axis=1,inplace=True)
```

```python
df.head()
```

```python
# First we make a Dendogram to find the no of cluster.
plt.figure(figsize=(20,20))
import scipy.cluster.hierarchy as sc
sc.dendrogram(sc.linkage(df,method='single',metric='euclidean'))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
ac = AgglomerativeClustering(n_clusters=2,linkage='single')
df['predict'] = ac.fit_predict(df)
```

```python
df.head()
```

```python
sns.pairplot(data=df,hue='predict')
plt.show()
```
