```python
#import the library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings("ignore")
```

```python
#load the datasets
df = pd.read_csv("iris_synthetic_data.csv")
df.head()
```

```python
# Drop the column of label
df.drop("label",axis=1,inplace=True)
```

```python
df.head()
```

```python
sns.pairplot(df)
```

```python
# Best no of clustering.
from sklearn.cluster import KMeans
wcss = [] 
for i in range(2,21):
    km = KMeans(n_clusters=i,init='k-means++')
    km.fit(df)
    wcss.append(km.inertia_)#wcss
```

```python
plt.plot([i for i in range(2,21)],wcss,marker='o')
plt.xlabel("no of cluster")
plt.xticks([i for i in range(2,21)])
plt.ylabel("wcss")
plt.grid(axis="x")
plt.show()
```

```python
kmn = KMeans(n_clusters=3)
df['predict']= kmn.fit_predict(df)
```

```python
df.head()
```

```python
sns.pairplot(data=df,hue='predict')
plt.show()
```
